```python
x = torch.tensor([0.7, 0.9, 10, -1, -5, 2, 4, 6.5])
```

## sigmoid 函数

### 使用方式1： 直接调用torch的sigmoid函数

```python
y = torch.sigmoid(x)
print(y)
```

### 使用方式2: 直接调用tensor的sigmoid函数

```python
y = x.sigmoid()

print(y)
```

### 使用方法3: nn模块中的Sigmoid 类 （常用）

```python
sigmoid = nn.Sigmoid() # 常用
y = sigmoid(x)
print(y)
```

## Tanh 函数


```python
print(torch.tanh(x))
print(x.tanh())
print(nn.Tanh()(x))
```

## ReLu 函数

```python
print(torch.relu(x))
print(x.relu())
print(nn.ReLU()(x))
```

### softmax函数

```python
print(torch.softmax(x, dim=-1))
print(x.softmax(dim=-1))
print(nn.Softmax(dim=-1)(x))
```

## 其他激活函数

```python
print(nn.LeakyReLU()(x))
print(nn.PReLU()(x))
print(nn.RReLU()(x))
print(nn.SiLU()(x))
print(nn.GELU()(x))
```
